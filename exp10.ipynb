## Pair plot

````python
sns.pairplot(
    crop_df,
    hue="label",
    vars=["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]
)
plt.show()
````

## Split data into train and test set

````python
X = crop_df[["N", "P", "K", "temperature", "humidity", "ph", "rainfall"]].values
y = crop_df["label"].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, train_size=0.9, random_state=1
)
````

## Train the Naïve Bayes Classifier

````python
clf = GaussianNB()
clf.fit(X_train, y_train)
````

## Test the model trained on test set

````python
prediction = clf.predict(X_test)
prediction[:10]
````

**OUTPUT**
```text
array(['rice', 'maize', 'cotton', 'banana', 'apple',
       'rice', 'coffee', 'jute', 'grapes', 'orange'], dtype=object)
```

## Perform Evaluation on Test set

````python
print(classification_report(y_test, prediction))
````

**OUTPUT**
```text
              precision    recall  f1-score   support

      accuracy                           0.99       220
     macro avg       0.99      0.99      0.99       220
  weighted avg       0.99      0.99      0.99       220
```

## Plot Confusion Matrix

````python
ConfusionMatrixDisplay.from_estimator(
    clf, X_test, y_test, xticks_rotation="vertical"
)
plt.show()
````

## Feeding new data to the classifier to predict the right crop

````python
N = 90
P = 42
K = 43
temperature = 20.8797
humidity = 82.0027
ph = 6.5030
rainfall = 202.9355

x = [[N, P, K, temperature, humidity, ph, rainfall]]

res = clf.predict(x)
predicted_crop = res[0]

print("The crop predicted is -->", predicted_crop)
````
